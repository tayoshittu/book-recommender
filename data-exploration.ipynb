```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("dylanjcastillo/7k-books-with-metadata")

print("Path to dataset files:", path)
```

```python
import pandas as pd
```

```python
books = pd.read_csv(f"{path}/books.csv")
#use pandas read csv method, change to f string to look at our file
```

```python
books
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#instantiate matplot lib plot
ax = plt.axes()
#make seaborn heatmap that documents all the missing values "isna()"
#and transpose() to make it nicer
sns.heatmap(books.isna().transpose(), cbar=False , ax=ax)
#add labels to the columns
plt.xlabel("Columns")
plt.ylabel("Missing Values")
plt.show()
```

```python
import numpy as np
#books where description is missing put a 1 in the column if not a 0
books["missing_description"] = np.where(books["description"].isna(), 1, 0)
#convert the year of the book to the age of the book
books["age_of_book"] = 2024 - books["published_year"]
```

```python
#return columns of interest
columns_of_interest = ["num_pages", "age_of_book", "missing_description", "average_rating"]
#correlation matrix directly off the pandas dataframe using just those columns
#using spearman because it's appropriate for binary values
#pearson is more appropriate for continuous values
correlation_matrix = books[columns_of_interest].corr(method="spearman")
#convert the correlation matrix to a seaborn heatmap
sns.set_theme(style= "white")
plt.figure(figsize=(8,10))
heatmap = sns.heatmap(correlation_matrix, annot=True, fmt=".2F", cmap="coolwarm",
                      cbar_kws={"label": "Spearman Correlation"})
heatmap.set_title("Correlation Heatmap")
plt.show()
```

```python
#weak correlations are closer to 0 while strong correlations are closer to 1 or -1
#based on these correletion efficient our system would have a good range of books of length and rating
```

```python
#find out how many missing values for description, num of pages, published year
books[(books["description"].isna())|
      (books["num_pages"].isna())|
      (books["published_year"].isna())
]
```

```python
#remove these rows that have missing values from the dataframe
books_missing = books[~(books["description"].isna()) &
      ~(books["num_pages"].isna()) &
      ~(books["average_rating"].isna()) &
      ~(books["published_year"].isna())
]
```

```python
books_missing
```

```python
#problem: there are alot of distinct values in categories column
#create a histogram to see the distribution of categories
#switch to chart view and change the type to bar, with x axis as categories and y axis as count
books_missing["categories"].value_counts().reset_index().sort_values("count", ascending=False)
```

```python
#this indicates a long tail problem with book categories because there are many categories with 1 book
#refine description categories by checking how long the descriptions are
#create a new variable "words in description" and take split the descriptions based on space and ask or length to give us the word count
books_missing["words_in_description"] = books_missing["description"].str.split().str.len()
```

```python
#switch to chart view, histogram and assign words in descriptions to both x and y axis(count) to see distribution
#from the view you see there isnt any break point of a clear group of short or long descriptions
books_missing
```

```python
# remove all the descriptions less than 25
books_missing_25_words = books_missing[books_missing["words_in_description"] >= 25]
books_missing_25_words
```

```python
# final cleaning steps
#create a new field called title and subtitle
#use np where with a clause "when subtitle is missing just use the title
#but when present, aggregate the subtitle and the title and convert to string, joining them using a colon
books_missing_25_words["title_and_subtitle"] = (
    np.where(books_missing_25_words["subtitle"].isna(), books_missing_25_words["title"],
             books_missing_25_words[["title", "subtitle"]].astype(str).agg(": ".join, axis=1))
)
```

```python
books_missing_25_words
```

```python
#create a new description column that has each description with a unique identifier important for building the vector database
books_missing_25_words["tagged_description"] = books_missing_25_words[["isbn13", "description"]].astype(str).agg(" ".join, axis=1)
```

```python
#drop columns that wont be needed
(
    books_missing_25_words
    .drop(["subtitle", "missing_description", "age_of_book", "words_in_description"], axis=1)
    .to_csv("books_cleaned.csv", index=False)
)
```
